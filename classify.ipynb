# Data Processing

```python
import tensorflow
```

```python
import pandas as pd
cancer = pd.read_csv("breastcancer.csv")
```

```python
cancer=cancer.drop(columns='Unnamed: 32')
cancer['diagnosis']=cancer['diagnosis'].replace({
    "M":1, "B":0
})
cancer.head()
```

```python
features=cancer.iloc[:,2:-1].dropna(axis=1)
features.head()
```

```python
X=features.to_numpy()
X=X.astype(float)
print(X.shape)
```

```python
target=cancer['diagnosis']
y=target.to_numpy()
y=y.astype(float)
print(y.shape)
```

```python
import torch
from torch.utils.data import TensorDataset
```

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_val, X_test, y_val, y_test = train_test_split(
    X_val,
    y_val,
    test_size=0.40,
    random_state=42,
    stratify=y_val
)
```

```python
datasetTrain = TensorDataset(torch.tensor(X_train),torch.tensor(y_train))
datasetVal = TensorDataset(torch.tensor(X_val),torch.tensor(y_val))
datasetTest = TensorDataset(torch.tensor(X_test),torch.tensor(y_test))
x1,y1 = datasetTrain[0]
print("Features:",x1) 
print("Target:",y1)
```

```python
from torch.utils.data import DataLoader

batch_size = 5 #receive 5 samples at a time
shuffle = True

trainLoader = DataLoader(datasetTrain, batch_size=batch_size,shuffle=shuffle)
valLoader = DataLoader(datasetVal, batch_size=batch_size,shuffle=shuffle)
testLoader = DataLoader(datasetTest, batch_size=batch_size,shuffle=shuffle)
```

```python
import torch.nn as nn

print(X.shape) #(569, 29)
model = nn.Sequential(
    nn.Linear(29,10),
    nn.ReLU(),
    nn.Linear(10,5),
    nn.ReLU(),
    nn.Linear(5,1),
    nn.Sigmoid()
)
x1=x1.float()
print(model(x1))
```

```python
import torch.optim as optim 
```

```python
criterion = nn.BCELoss() #loss function to be used
optimizer = optim.SGD(model.parameters(),lr=0.001)
```

```python

for epoch in range(13):
    for data in trainLoader: #training loop for training data
        optimizer.zero_grad() #set original gradient to 0 
        feature,target = data
        feature=feature.float()
        target=target.float()
        pred=model(feature).squeeze(1)
        loss = criterion(pred,target)
        train_loss = loss.item()
        loss.backward() #gradients are calculated
        optimizer.step() #weights are updated
    mean_train_loss = train_loss/len(trainLoader)
    model.eval() #switching to models evaluation mode
    validation_loss=0
    with torch.no_grad(): #disable gradients
        for inputs, labels in valLoader:
            inputs=inputs.float()
            labels=labels.float()
            outputs = model(inputs).squeeze(1)
            loss = criterion(outputs,labels)
            validation_loss = loss.item()

        mean_loss = validation_loss/len(valLoader)
    print("Train Loss:",mean_train_loss,"Validation Loss:",mean_loss)
    model.train()
```

```python
import torchmetrics

metric = torchmetrics.Accuracy(task="binary")
with torch.no_grad():
    for features, labels in testLoader:
        features=features.float()
        labels=labels.float()
        outputs = model(features).squeeze(1)
        metric.update(outputs,labels)

accuracy= metric.compute()
print(accuracy)
metric.reset()
```
